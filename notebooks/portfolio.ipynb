# 投资组合分析
持仓分布、净值曲线和风险指标

```python
import sys, os
sys.path.insert(0, os.path.dirname(os.path.abspath(".")))

import pandas as pd
import plotly.express as px
from helpers import get_portfolio, get_episode_summaries

try:
    portfolio = get_portfolio()
    episodes = get_episode_summaries()
    has_data = bool(portfolio) or len(episodes) > 0
except Exception as e:
    portfolio = {}
    episodes = pd.DataFrame()
    has_data = False
```

## 当前持仓分布

```python
if has_data and portfolio:
    positions = portfolio.get("positions", {})
    if positions:
        df = pd.DataFrame(list(positions.items()), columns=["stock", "shares"])
        fig = px.pie(df, values="shares", names="stock", title="当前持仓分布")
        fig.show()
    else:
        print("当前无持仓")
else:
    print("暂无投资组合数据")
```

## 历史净值曲线

```python
if len(episodes) > 0:
    fig = px.line(episodes, x=episodes.index, y="final_nav", title="历史NAV")
    fig.update_layout(xaxis_title="轮次", yaxis_title="NAV")
    fig.show()
```

## 风险指标

```python
if len(episodes) > 0:
    fig = px.bar(episodes, x=episodes.index, y="sharpe", title="Sharpe对比")
    fig.update_layout(xaxis_title="轮次", yaxis_title="Sharpe")
    fig.show()

    fig2 = px.bar(episodes, x=episodes.index, y="max_drawdown", title="最大回撤对比")
    fig2.update_layout(xaxis_title="轮次", yaxis_title="最大回撤")
    fig2.show()
```
