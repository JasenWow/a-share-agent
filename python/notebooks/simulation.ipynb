# 策略进化模拟

可视化 Meta-Agent 策略进化过程

```python
import sys
import os

sys.path.insert(0, os.path.dirname(os.path.abspath(".")))

import pandas as pd
import plotly.express as px
from helpers import get_experiments, get_best_strategies, get_episode_summaries

# Try to fetch data, handle empty gracefully
try:
    experiments = get_experiments(limit=100)
    best_strategies = get_best_strategies(top_k=5)
    episodes = get_episode_summaries()
    has_data = len(experiments) > 0
except Exception:
    experiments = pd.DataFrame()
    best_strategies = pd.DataFrame()
    episodes = pd.DataFrame()
    has_data = False
```

```python
if not has_data:
    print("暂无模拟数据，请先运行 Meta-Agent 进化循环")
else:
    print(f"加载了 {len(experiments)} 条实验记录")
```

```python
if has_data and len(experiments) > 0:
    # Extract final_nav from result JSON
    df = experiments.copy()
    if "result" in df.columns:
        df["final_nav"] = df["result"].apply(lambda x: x.get("final_nav", 0) if isinstance(x, dict) else 0)
        df["sharpe"] = df["result"].apply(lambda x: x.get("sharpe", 0) if isinstance(x, dict) else 0)

    fig = px.line(df, x=df.index, y="final_nav", title="策略性能随迭代变化")
    fig.update_layout(xaxis_title="迭代", yaxis_title="最终NAV")
    fig.show()
```

```python
if has_data and "sharpe" in df.columns and len(df) > 0:
    fig = px.histogram(df, x="sharpe", nbins=20, title="Sharpe比率分布")
    fig.update_layout(xaxis_title="Sharpe", yaxis_title="频数")
    fig.show()
```

```python
if has_data and len(best_strategies) > 0:
    display(best_strategies[["id", "final_nav", "sharpe"]].head())
```

```python
if has_data and len(episodes) > 0:
    fig = px.bar(episodes, x=episodes.index, y="final_nav", title="各轮次最终NAV对比")
    fig.update_layout(xaxis_title="轮次", yaxis_title="最终NAV")
    fig.show()
```
